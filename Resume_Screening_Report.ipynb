## 1. Introduction

The AI-Based Resume Screening and Candidate Analysis System is a Python-based application that analyzes resumes against a given job description. It extracts candidate skills, identifies matching and missing skills, calculates skill matching scores, and compares candidates using TF-IDF text similarity.

The application provides both single-resume and multiple-resume analysis and generates downloadable CSV reports.

## 2. Objectives

- Extract text from PDF, DOCX and TXT resumes.
- Identify candidate skills.
- Compare candidate skills with job requirements.
- Calculate skill matching scores.
- Measure text similarity using TF-IDF and cosine similarity.
- Rank multiple candidates based on similarity scores.
- Generate candidate analysis results and CSV reports.

## 5. Methodology

1. The user enters a job description.
2. The user uploads one or more resumes.
3. Resume text is extracted.
4. Text is cleaned and relevant skills are identified.
5. Candidate skills are compared with required job skills.
6. A skill matching score is calculated.
7. TF-IDF converts text into numerical vectors.
8. Cosine similarity measures resume-to-job-description similarity.
9. Results are displayed and can be downloaded as a CSV report.

### Skill Matching Formula

Skill Matching Score = (Number of Matched Skills / Total Required Skills) × 100

### TF-IDF and Cosine Similarity

TF-IDF represents text using numerical features based on term frequency and inverse document frequency. Cosine similarity compares the resulting vectors to estimate textual similarity.

## 8. Actual CSV Output Analysis

The application generated a combined candidate report containing identified skills, matched skills, missing skills, extra skills, skill counts, matching scores, similarity scores and candidate summaries.

The following code reads the actual CSV generated by the application.


```python

import csv
from pathlib import Path

csv_path = Path("outputs/multiple_candidate_results.csv")

if not csv_path.exists():
    raise FileNotFoundError(f"Output CSV not found: {csv_path}")

with csv_path.open("r", encoding="utf-8", newline="") as file:
    results = list(csv.DictReader(file))

print("Actual Multiple-Candidate Analysis Results")
print("------------------------------------------")

for candidate in results:
    print("\nCandidate:", candidate["Candidate"])
    print("Matched skills:", candidate["Total Matched Skills"])
    print("Missing skills:", candidate["Total Missing Skills"])
    print("Skill matching score:", candidate["Skill Matching Score"], "%")
    print("TF-IDF similarity:", candidate["TF-IDF Similarity Score"], "%")
    print("Candidate summary:", candidate["Candidate Summary"])
```

## 11. Limitations

- Skill extraction depends on a predefined keyword and alias list.
- TF-IDF similarity measures textual similarity and does not establish a candidate's actual ability.
- The system does not train a supervised classification model using labelled candidate data.
- Automated scores should be treated as estimates, not final recruitment decisions.
- Human review is necessary for fair and informed candidate evaluation.
